```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader
from torchvision import datasets, transforms
import matplotlib.pyplot as plt

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
device
```

TẢI DỮ LIỆU


```python
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.1307,), (0.3081,))
])

train_ds = datasets.MNIST(root="./data", train=True, download=True, transform=transform)
test_ds  = datasets.MNIST(root="./data", train=False, download=True, transform=transform)

train_loader = DataLoader(train_ds, batch_size=64, shuffle=True)
test_loader  = DataLoader(test_ds, batch_size=64, shuffle=False)
```

HÀm train & evalute


```python
def train_model(model, train_loader, test_loader, epochs=10, lr=1e-3, wd=0.0):
    model.to(device)
    opt = torch.optim.Adam(model.parameters(), lr=lr, weight_decay=wd)
    crit = nn.CrossEntropyLoss()

    tr_loss, te_loss, tr_acc, te_acc = [], [], [], []

    for ep in range(epochs):
        model.train()
        correct = total = 0
        loss_sum = 0

        for x, y in train_loader:
            x, y = x.to(device), y.to(device)
            opt.zero_grad()
            out = model(x)
            loss = crit(out, y)
            loss.backward()
            opt.step()

            loss_sum += loss.item() * x.size(0)
            correct += (out.argmax(1) == y).sum().item()
            total += y.size(0)

        tr_loss.append(loss_sum / total)
        tr_acc.append(correct / total)

        # evaluate
        model.eval()
        correct = total = 0
        loss_sum = 0
        with torch.no_grad():
            for x, y in test_loader:
                x, y = x.to(device), y.to(device)
                out = model(x)
                loss = crit(out, y)
                loss_sum += loss.item() * x.size(0)
                correct += (out.argmax(1) == y).sum().item()
                total += y.size(0)

        te_loss.append(loss_sum / total)
        te_acc.append(correct / total)

        print(f"Epoch {ep+1}/{epochs} | Train acc {tr_acc[-1]:.4f} | Test acc {te_acc[-1]:.4f}")

    return tr_loss, te_loss, tr_acc, te_acc
```

MLP với reLU + Regulaiation

```python
class MLP_ReLU_Reg(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(784, 512)
        self.fc2 = nn.Linear(512, 256)
        self.fc3 = nn.Linear(256, 128)
        self.fc4 = nn.Linear(128, 10)
        self.dp = nn.Dropout(0.3)

    def forward(self, x):
        x = x.view(x.size(0), -1)
        x = F.relu(self.fc1(x)); x = self.dp(x)
        x = F.relu(self.fc2(x)); x = self.dp(x)
        x = F.relu(self.fc3(x))
        return self.fc4(x)
```

Train model

```python
mlp_reg = MLP_ReLU_Reg()
tr_loss1, te_loss1, tr_acc1, te_acc1 = train_model(
    mlp_reg, train_loader, test_loader, epochs=15, lr=1e-3, wd=1e-4
)
```

```python
plt.plot(tr_loss1, label="train")
plt.plot(te_loss1, label="test")
plt.legend()
plt.title("MLP ReLU + Regularization - Loss")
plt.show()

plt.plot(tr_acc1, label="train")
plt.plot(te_acc1, label="test")
plt.legend()
plt.title("MLP ReLU + Regularization - Accuracy")
plt.show()
```

MLP không regularization + Sigmoid version

```python
class MLP_ReLU_NoReg(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(784, 512)
        self.fc2 = nn.Linear(512, 256)
        self.fc3 = nn.Linear(256, 128)
        self.fc4 = nn.Linear(128, 10)

    def forward(self, x):
        x = x.view(x.size(0), -1)
        x = F.relu(self.fc1(x))
        x = F.relu(self.fc2(x))
        x = F.relu(self.fc3(x))
        return self.fc4(x)

class MLP_Sigmoid(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(784, 512)
        self.fc2 = nn.Linear(512, 256)
        self.fc3 = nn.Linear(256, 128)
        self.fc4 = nn.Linear(128, 10)
        self.dp = nn.Dropout(0.3)

    def forward(self, x):
        x = x.view(x.size(0), -1)
        x = torch.sigmoid(self.fc1(x)); x = self.dp(x)
        x = torch.sigmoid(self.fc2(x)); x = self.dp(x)
        x = torch.sigmoid(self.fc3(x))
        return self.fc4(x)
```

Train NoReg + Sigmoid

```python
mlp_noreg = MLP_ReLU_NoReg()
tr_loss2, te_loss2, tr_acc2, te_acc2 = train_model(mlp_noreg, train_loader, test_loader, epochs=15)
```

```python
mlp_sig = MLP_Sigmoid()
tr_loss3, te_loss3, tr_acc3, te_acc3 = train_model(mlp_sig, train_loader, test_loader, epochs=15, wd=1e-4)
```

```python
plt.figure(figsize=(12, 4))

# Loss plot
plt.subplot(1, 2, 1)
plt.plot(te_loss1, 'b-', label='ReLU + Reg')
plt.plot(te_loss2, 'g-', label='ReLU No Reg')
plt.plot(te_loss3, 'r-', label='Sigmoid')
plt.xlabel('Epoch')
plt.ylabel('Test Loss')
plt.title('Loss Comparison')
plt.legend()
plt.grid(True)

# Accuracy plot
plt.subplot(1, 2, 2)
plt.plot(te_acc1, 'b-', label='ReLU + Reg')
plt.plot(te_acc2, 'g-', label='ReLU No Reg')
plt.plot(te_acc3, 'r-', label='Sigmoid')
plt.xlabel('Epoch')
plt.ylabel('Test Accuracy (%)')
plt.title('Accuracy Comparison')
plt.legend()
plt.grid(True)

plt.suptitle('Model Comparison: Regularization & Activation Functions', fontsize=12)
plt.tight_layout()
plt.show()
```

CNN

```python
class SimpleCNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.conv1 = nn.Conv2d(1, 16, 3, padding=1)
        self.conv2 = nn.Conv2d(16, 32, 3, padding=1)
        self.pool = nn.MaxPool2d(2,2)
        self.fc1 = nn.Linear(32*7*7, 128)
        self.fc2 = nn.Linear(128, 10)

    def forward(self, x):
        x = F.relu(self.conv1(x))
        x = self.pool(x)
        x = F.relu(self.conv2(x))
        x = self.pool(x)
        x = x.view(x.size(0), -1)
        x = F.relu(self.fc1(x))
        return self.fc2(x)
```

```python
cnn = SimpleCNN()
cnn_tr_loss, cnn_te_loss, cnn_tr_acc, cnn_te_acc = train_model(
    cnn, train_loader, test_loader, epochs=5, lr=1e-3, wd=1e-4
)
```

Visualize activation

```python
images, labels = next(iter(test_loader))
images = images.to(device)

acts = {}

def hook(name):
    def fn(m, inp, out):
        acts[name] = out.detach().cpu()
    return fn

h1 = cnn.conv1.register_forward_hook(hook("c1"))
h2 = cnn.conv2.register_forward_hook(hook("c2"))

cnn.eval()
with torch.no_grad():
    _ = cnn(images)

h1.remove()
h2.remove()

act1 = acts["c1"]
act2 = acts["c2"]

plt.figure(figsize=(10,4))
for i in range(6):
    plt.subplot(2,3,i+1)
    plt.imshow(act1[0, i], cmap="gray")
    plt.axis("off")
plt.suptitle("Conv1 Feature Maps")
plt.show()
```
